# Simulación de datos para empresa Retail mediana en México
## Objetivo
Generar un dataset que simule el comporrtamiento afectado por estacionalidad y que muestre una tendencia creible para un modelo de forecast

```python
import pandas as pd
import numpy as np
#Parametros
fecha_inicio = "2024-01-01"
fecha_final= "2025-12-31"
n_tiendas= 5
n_productos= 50
#Rango de fechas
fechas=pd.date_range(start=fecha_inicio, end=fecha_final, freq="D")
#Sucursales y productos
sucursales= [f"Sucursal_{i}" for i in range(1, n_tiendas + 1)]
productos = [f"Producto_{i}" for i in range(1, n_productos + 1)]
# Estructura
df=pd.MultiIndex.from_product(
    [fechas, sucursales, productos],
    names=["fecha", "tienda_id", "producto_id"]
).to_frame(index=False)
df.head
```
